```python
def create_at_rec(n):
    if n == 1:
        return {"0",'1'}
    elif n == 2:
        return {'00','01','10','11'}
    elif n == 3:
        return {'001','010','011','100','101','110','111'}
    result = set()
    base1 = create_at_rec(n-1)
    base2 = [x for x in create_at_rec(n-2) if not x.endswith('00')]
    base3 = [x for x in create_at_rec(n-3) if x.endswith('1')]
    result = result.union(base1).union(base2).union(base3).union(
        [x+'1' for x in base1] + [x+'01' for x in base2] + [x+'001' for x in base3]
        )
    return result
[x for x in create_at_rec(4) if len(x)==4]
```

```python
def append_all(A,x):
    if len(A) == 0: return []
    else: return [A[0] + x] + append_all(A[1:],x)

def get_end_with(A, x):
    if len(A) == 0: return []
    elif len(A) == 1: return [A[0]] if A[0].endswith(x) else []
    else: #len(A) >= 2
        result = get_end_with([A[0]],x)
        return result + get_end_with(A[1:], x)

def create_at_functional(n):
    if n == 0: return []
    elif n == 1: return ['0','1']
    elif n == 2: return ['00','01','10','11']
    elif n == 3: return ['001','010','011','100','101','110','111']
    else:
        base = create_at_functional(n-2)
        base_1 = get_end_with(base, '1')
        base_00 = get_end_with(base, '00')
        base_10 = get_end_with(base, '10')
        result_1 = append_all(base_1, '10') + append_all(base_1, '11') + append_all(base_1, '00') +append_all(base_1, '01') 
        result_00 = append_all(base_00, '10') + append_all(base_00, '11')
        result_10 = append_all(base_10, '10') + append_all(base_10, '11') + append_all(base_10, '01')
        return result_1 + result_00 + result_10

create_at_functional(6)
```

```python
class BalanceTree:

    def __init__(self):
        self.sum_weight = None
        self.LeftTree = None
        self.RightTree = None
    
    def setLeftTree(self,LT):
        self.LeftTree = LT
        if self.RightTree:
            self.sum_weight = self.LeftTree.sum_weight + self.RightTree.sum_weight
        else:
            self.sum_weight = self.LeftTree.sum_weight

    def setRightTree(self,RT):
        self.RightTree = RT
        if self.LeftTree:
            self.sum_weight = self.LeftTree.sum_weight + self.RightTree.sum_weight
        else:
            self.sum_weight = self.RightTree.sum_weight

    def checkBalance(self):
        left_balance = self.LeftTree.checkBalance()
        right_balance = self.RightTree.checkBalance()
        if (left_balance and right_balance):
            return self.LeftTree.sum_weight == self.RightTree.sum_weight
        else:
            return False    
```
