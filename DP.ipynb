$MC(n) = 1 + \min\{MC(n-1), MC(n-2), MC(n-8), MC(n-10)\}$

$MC(1), MC(2), MC(8), MC(10) = 1$

```python
# besic recursion

def MC(n,coin_sys = [1,2,8,10]):
    if n in coin_sys:
        return 1
    elif n <= 0:
        return None
    else:
        print(f"n = {n}")
        all_choice = [MC(n - x, coin_sys) for x in coin_sys]
        return 1 + min([x for x in all_choice if x is not None])
    
MC(50)
```

```python
# Top-down recursion: Memoization ---- แก้ปัญหา overlapping subproblems ไม่ได้แก้ stack recursion depth

SOLUTION_DICT = {1:1, 2:1, 8:1, 10:1}
def MC_top_down(n,coin_sys = [1,2,8,10]):
    global SOLUTION_DICT
    if n in SOLUTION_DICT.keys():
        return SOLUTION_DICT[n]
    elif n <= 0:
        return None
    else:
        all_choice = [MC_top_down(n - x, coin_sys) for x in coin_sys]
        result = 1 + min([x for x in all_choice if x is not None])
        SOLUTION_DICT[n] = result
        return result
MC_top_down(25818)
```

$MC(n) = 1 + \min\{MC(n-1), MC(n-2), MC(n-8), MC(n-10)\}$

```python
# BOTTOM UP approach ---- แก้ปัญหา recursion depth
# เติมตารางไล่ขึ้นจากน้อยไปมาก จนกว่าจะได้ขั้นที่เราอยากได้
SOLUTION_DICT = {1:1, 2:1, 8:1, 10:1}
def MC_bottom_up(n,coin_sys = [1,2,8,10]):
    global SOLUTION_DICT
    if n in SOLUTION_DICT.keys():
        return SOLUTION_DICT[n]
    for i in range(3,n+1):
        all_result = [SOLUTION_DICT.get(i-x,None) for x in coin_sys]
        SOLUTION_DICT[i] = 1 + min([x for x in all_result if x is not None])
    return SOLUTION_DICT[n]
MC_bottom_up(25818)
```
