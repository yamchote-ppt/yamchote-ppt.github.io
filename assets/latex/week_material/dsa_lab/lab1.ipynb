## นับ Operation

```python
n = 10
print(f"รับ n = {n}")
num_op = 0
##########IMPLEMENT HERE##############
x = 0
num_op += 1 # num_op = num_op + 1
for i in range(1,n+1):
    num_op += 1 # นับ i = ...
    x = x + 2
    num_op += 2 # นับ x + 2 และ นับ x = ...
print(f"ได้ x = {x}")
num_op += 1 # นับการ return
######################################
print(f"จำนวนครั้งการดำเนินการมีค่าเท่ากับ {num_op}")
```

## ทำการทดลองหลาย `n` เพื่อ plot graph

ดูค่า `num_op` ของแต่ละ `n`

```python
result = []
for n in range(1, 20+1):
    num_op = 0
    ##########IMPLEMENT HERE##############
    x = 0
    num_op += 1 # num_op = num_op + 1
    for i in range(1,n+1):
        num_op += 1 # นับ i = ...
        x = x + 2
        num_op += 2 # นับ x + 2 และ นับ x = ...
    num_op += 1 # นับการ return
    result.append(num_op)
    ######################################
import matplotlib.pyplot as plt
plt.plot(result)
plt.xlabel("n")
plt.ylabel("num operator")
```

# FindFirstZero

```python
A = [1,2,0,5,6,7,8]
n = len(A) # หาว่า A มีสมาชิกกี่ตัว

###### IMPLEMENT HERE #######
found = False # เป็นตัวบอกว่าหาเจอหรือยัง (ตั้งเป็นเท็จไว้ก่อนเพราะยังไม่ได้หา)
for i in range(n): # i = 0, ..., n-1
    if A[i] == 0:
        found = True
        break

# หลุดออกจาก for loop ได้ 2 กรณี
# (1) เจอเลข 0------------------------ found = True
# (2) วิ่งหมดครบทุกรอบ แต่ก็ไม่เจอ 0------- found = False
if found:
    print(f"เจอเลข 0 ที่ index ที่ {i}")
else:
    print("ไม่เจอ")
#############################
```

```python
A = [1,2,5,6,7,8,0]
n = len(A) # หาว่า A มีสมาชิกกี่ตัว
num_op = 0
found = False # เป็นตัวบอกว่าหาเจอหรือยัง (ตั้งเป็นเท็จไว้ก่อนเพราะยังไม่ได้หา)
for i in range(n): # i = 0, ..., n-1
    num_op += 1 # จากการกำหนดค่า i = ...
    num_op += 1 # เข้าถึง index ที่ i ของ A
    num_op += 1 # เปรียบเทียบกับ 0 ว่าเท่าหรือไม่
    if A[i] == 0:
        found = True
        break
num_op += 1 # การ return
if found:
    print(f"เจอเลข 0 ที่ index ที่ {i} | num_op = {num_op}")
else:
    print(f"ไม่เจอ | num_op = {num_op}")
```
